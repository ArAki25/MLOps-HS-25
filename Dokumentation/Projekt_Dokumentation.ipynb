### Ziele und Aufgabenstellung
#### Das Projektkonzept
Simap.ch ist die offizielle Schweizer Publikationsplattform für öffentliche Beschaffungen.  Diese Plattform hat eine zentrale Funktion. Sie ermöglicht Unternehmen zwei wichtige Dinge. Erstens: Sie gewährt ihnen Einsicht in Ausschreibungen. Zweitens: Sie erlaubt ihnen die Abgabe von Offerten. Somit trägt sie zu mehr Transparenz und Fairness im öffentlichen Beschaffungswesen bei.

#### Zielsetzung
Unsere Grundidee ist die Entwicklung einer künstlichen Intelligenz (KI), die als digitaler Berater eingesetzt werden kann. Diese KI soll Unternehmen, die sich über Simap.ch bewerben, eine strategische Analyse bieten. Genau gesagt wollen wir folgende Vorteile schaffen:
- Projektbewertung: Identifizierung von Projekten, die für die jeweilige Firma sinnvoll und erfolgsversprechend ist.
- Zuschlagsanalyse: Bewertung der Erfolgschancen basierend auf hstorischen Zuschlägen und Mustern.

### Methodik, Vorgehensweise und Prozeduren
#### API-Anbindung und Datenerstellung
Damit wir unsere Klassifikationsaufgabe durchführen können, brauchen wir strukturierte historische Daten zu Ausschreibungen und Zuschlägen von simap.ch. Weil simap.ch im Web-Interface keine offenen Download-Funktionen anbietet, haben wir entschieden, die offizielle REST-API von simap.ch zu verwenden. Damit beziehen wir die Daten programmgesteuert und exportieren sie in ein geeignetes Analyseformat (CSV).

#### API von simap.ch einrichten
Schritte, die wir durchgeführt haben:

- Registrierung auf simap.ch als Organisation, um Zugriff auf die neue  Plattform (ab 2024) zu erhalten.

- Anfrage nach API-Zugang über die Administratoren von simap.ch (die API befindet sich aktuell im produktiven Betrieb für Systemintegrationen).

- Nach Freischaltung erhielten wir:

    einen API-Key (Bearer Token) zur Authentifizierung,

    die Basis-URL der Schnittstelle:

    die API-Dokumentation mit verfügbaren Endpunkten (z. B. /notices, /tenders, /awards).

#### Datenexport

Um die Daten weiter in Python oder in einem Notebook analysieren zu können, haben wir die abgerufenen JSON-Daten in eine CSV-Datei umgewandelt.

#### Auswahl relevanter Attribute
Nachdem wir die Rohdaten erfolgreich über die API von simap.ch beziehen und in CSV-Format speichern konnten, bestand der nächste Schritt darin, relevante Attribute für unsere Klassifikationsmodelle auszuwählen. Ziel war es, die Daten so aufzubereiten, dass sie von Machine-Learning-Algorithmen verarbeitet werden können.

##### Zielattribut
Für unsere Klassifikationsmodelle benötigten wir eine konkrete Zielvariable.
Wir haben uns für eine binäre Klassifikation entschieden:
- ist_interessant
- nicht_interessant

Diese Entscheidung wurde getroffen, weil wir Unternehmen ermöglichen wollen, schnell zu erkennen, ob eine Ausschreibung für sie potenziell relevant ist. Die Einteilung basiert auf verschiedenen Kriterien, z. B.:
- Höhe des Auftragswerts
- Projekttyp
- Region
- textliche Beschreibung des Projekts

#### Datenbank-Integration mit Supabase

Für die Speicherung, Verwaltung und Bereitstellung der strukturierten Projektdaten haben wir Supabase als Datenbanklösung eingesetzt.
Supabase bietet eine PostgreSQL-basierte Cloud-Datenbank mit integrierten Authentifizierungs- und API-Funktionen. Dies ermöglichte uns:

- eine sichere Speicherung der Ausschreibungsdaten
- eine klare Strukturierung der Tabellen (Ausschreibungen, Firmen, Klassifikationsergebnisse)
- eine einfache Anbindung der Webapplikation über REST-Endpunkte
- role-based access für einzelne Benutzer

Supabase fungierte als zentrale Datendrehscheibe, welche sowohl von der Python-Webapplikation als auch vom ML-Modell angesprochen wird.

Parallel zur Datenbeschaffung und Modellentwicklung haben wir eine Webanwendung entwickelt, die Unternehmen den praktischen Zugriff auf die Analyseergebnisse ermöglicht.

Die Webapplikation wurde wie folgt aufgebaut:

#### Login-System

Um den Zugang nur berechtigten Unternehmen zu erlauben, implementierten wir ein Login-System.
Für die Testversion nutzten wir eine Musterfirma, um typische Benutzerinteraktionen zu simulieren.

Das Login prüft:

Benutzername & Passwort

Zugehörigkeit zur Firma

Zugriffsdaten aus Supabase

Erst nach erfolgreicher Authentifizierung wird der Benutzer zum Dashboard weitergeleitet.

#### Dashboard

Nach dem Login gelangt der Nutzer auf ein Dashboard, das einen umfassenden Überblick über relevante Informationen bietet.
Das Dashboard umfasst folgende Bereiche:

Tägliche Empfehlungen:
Eine dynamische Liste an Ausschreibungen, die auf Basis des trainierten Klassifikationsmodells für die Firma als relevant eingestuft wurden.

Laufende Aufträge:
Ausschreibungen, für die das Unternehmen eine Offerte geplant oder eingereicht hat.

Abgeschlossene Aufträge:
Historische Ausschreibungen, bei denen bereits ein Zuschlag vergeben wurde.

Dieses Dashboard wird automatisch aus Supabase gespeist und aktualisiert.
