### API-Anbindung und Datenerstellung
Damit wir unsere Klassifikationsaufgabe durchführen können, brauchen wir strukturierte historische Daten zu Ausschreibungen und Zuschlägen von simap.ch. Weil simap.ch im Web-Interface keine offenen Download-Funktionen anbietet, haben wir entschieden, die offizielle REST-API von simap.ch zu verwenden. Damit beziehen wir die Daten programmgesteuert und exportieren sie in ein geeignetes Analyseformat (CSV).

### API von simap.ch einrichten
Schritte, die wir durchgeführt haben:

- Registrierung auf simap.ch als Organisation, um Zugriff auf die neue  Plattform (ab 2024) zu erhalten.

- Anfrage nach API-Zugang über die Administratoren von simap.ch (die API befindet sich aktuell im produktiven Betrieb für Systemintegrationen).

- Nach Freischaltung erhielten wir:

    einen API-Key (Bearer Token) zur Authentifizierung,

    die Basis-URL der Schnittstelle:

    die API-Dokumentation mit verfügbaren Endpunkten (z. B. /notices, /tenders, /awards).

### Datenexport

Um die Daten weiter in Python oder in einem Notebook analysieren zu können, haben wir die abgerufenen JSON-Daten in eine CSV-Datei umgewandelt.

#### Auswahl relevanter Attribute
Nachdem wir die Rohdaten erfolgreich über die API von simap.ch beziehen und in CSV-Format speichern konnten, bestand der nächste Schritt darin, relevante Attribute für unsere Klassifikationsmodelle auszuwählen. Ziel war es, die Daten so aufzubereiten, dass sie von Machine-Learning-Algorithmen verarbeitet werden können.

##### Zielattribut
Für unsere Klassifikationsmodelle benötigten wir eine konkrete Zielvariable.
Wir haben uns für eine binäre Klassifikation entschieden:
- ist_interessant
- nicht_interessant

Diese Entscheidung wurde getroffen, weil wir Unternehmen ermöglichen wollen, schnell zu erkennen, ob eine Ausschreibung für sie potenziell relevant ist. Die Einteilung basiert auf verschiedenen Kriterien, z. B.:
- Höhe des Auftragswerts
- Projekttyp
- Region
- textliche Beschreibung des Projekts

### Datenbank-Integration mit Supabase

Für die Speicherung, Verwaltung und Bereitstellung der strukturierten Projektdaten haben wir Supabase als Datenbanklösung eingesetzt.
Supabase bietet eine PostgreSQL-basierte Cloud-Datenbank mit integrierten Authentifizierungs- und API-Funktionen. Dies ermöglichte uns:

- eine sichere Speicherung der Ausschreibungsdaten
- eine klare Strukturierung der Tabellen (Ausschreibungen, Firmen, Klassifikationsergebnisse)
- eine einfache Anbindung der Webapplikation über REST-Endpunkte
- role-based access für einzelne Benutzer

Supabase fungierte als zentrale Datendrehscheibe, welche sowohl von der Python-Webapplikation als auch vom ML-Modell angesprochen wird.









## Ergebnisse
### Webapplikation: SAJF Strategies

Wir haben eine B2B Web-Applikation entwickelt, die Schweizer Unternehmen hilft, öffentliche Ausschreibungen von simap.ch zu finden. Die Plattform bietet eine kostenlose Version mit Such- und Filterfunktionen sowie eine Pro-Version mit ML-basierten Empfehlungen.


#### Verwendete Technologien

Für das Backend haben wir Python Flask verwendet, da es einfach zu lernen ist und sich gut für kleinere Webanwendungen eignet. Als Datenbank nutzen wir Supabase, eine gehostete PostgreSQL-Datenbank mit eingebauter API. Das Frontend wurde mit HTML, CSS und JavaScript erstellt, ohne zusätzliche Frameworks. Die Website wird auf Render.com gehostet, da es kostenloses Hosting für kleine Projekte anbietet und sich direkt mit GitHub verbinden lässt.


#### Struktur

Die Struktur der Webseite besteht aus mehreren Dateien. Die Hauptdatei ist app.py, welche den Flask-Server startet und alle Routen definiert. Die Datei supabase_client.py enthält alle Funktionen für die Datenbankverbindung und Abfragen. In der requirements.txt stehen alle Python-Pakete, die installiert werden müssen. Im Ordner templates befinden sich alle HTML-Seiten wie die Hauptseite, Login-Seite und das Pro-Dashboard.

#### Supabase-Einrichtung

Zuerst haben wir ein Projekt auf supabase.com erstellt. Dabei erhält man eine Project URL und einen API Key, die später für die Verbindung benötigt werden.

In Supabase haben wir mehrere Tabellen erstellt. Die Tabelle projects_website enthält alle Ausschreibungen von simap.ch. Die Tabelle pro_users speichert die Login-Daten der Pro-Benutzer. Für jede Beispielfirma gibt es eine eigene Tabelle mit ML-Predictions, zum Beispiel BauUnternehmenAG, ITAG und BeschaffungsAG.

Wichtig war, dass wir die Row Level Security für alle Tabellen deaktiviert haben, damit die Website ohne Authentifizierung auf die Daten zugreifen kann.


#### Render.com Deployment

Für das Hosting haben wir Render.com verwendet. Zuerst haben wir uns dort angemeldet und einen neuen Web Service erstellt. Dann haben wir unser GitHub-Repository verbunden, sodass Render automatisch den Code laden kann.

Bei den Build-Einstellungen haben wir Python 3 als Environment gewählt. Als Build Command haben wir pip install -r requirements.txt angegeben, damit alle Abhängigkeiten installiert werden. Als Start Command verwenden wir gunicorn app:app, welches den Flask-Server startet.

Anschliessend haben wir die Umgebungsvariablen gesetzt. Diese sind SUPABASE_URL mit der Supabase-Projektadresse, SUPABASE_KEY mit dem API-Schlüssel und SECRET_KEY für die Flask-Sessions.

Nach dem Einrichten wird bei jedem Git Push automatisch eine neue Version deployed. Der Prozess dauert etwa zwei bis drei Minuten.


#### Verbindung der Komponenten

Die Website verbindet drei Hauptkomponenten miteinander. Die Ausschreibungsdaten kommen ursprünglich von simap.ch und werden in der Supabase-Datenbank gespeichert. Das vortrainierte ML-Modell, die vorhersagen, welche Aufträge für welche Firma relevant sind (Diese Predictions werden ebenfalls in Supabase gespeichert.

Die Flask-Website läuft auf Render.com und verbindet sich mit Supabase über API-Aufrufe. Wenn ein Benutzer die Seite öffnet, holt die Website die Daten aus der Datenbank und zeigt sie an. Bei der Pro-Version werden zusätzlich die ML-Predictions aus den firmenspezifischen Tabellen geladen.

#### Website-Features

Die kostenlose Version bietet eine Suchfunktion, mit der man nach Keywords in Titel und Beschreibung suchen kann. Es gibt Filter für alle 26 Kantone, Verfahrenstypen wie Offen oder Selektiv, und Auftragstypen wie Bau, Dienstleistung oder Lieferung. Die Aufträge werden als Karten angezeigt, die man anklicken kann um mehr Details zu sehen. Jeder Auftrag hat einen direkten Link zur offiziellen Ausschreibung auf simap.ch.

Die Pro-Version ermöglicht es Firmen, sich einzuloggen und personalisierte Empfehlungen zu sehen. Diese basieren auf den ML-Modellen und zeigen eine Match-Rate an, die angibt wie relevant ein Auftrag für die Firma ist. Die Aufträge sind nach dieser Relevanz sortiert, sodass die passendsten zuerst erscheinen.

Wenn die Seite einen Application Error zeigt, sollte man die Logs in Render prüfen. Meistens fehlen die Umgebungsvariablen oder es gibt einen Fehler im Code.
