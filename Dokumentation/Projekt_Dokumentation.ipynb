## Das Projektkonzept
Simap.ch ist die offizielle Schweizer Publikationsplattform für öffentliche Beschaffungen.  Diese Plattform hat eine zentrale Funktion. Sie ermöglicht Unternehmen zwei wichtige Dinge. Erstens: Sie gewährt ihnen Einsicht in Ausschreibungen. Zweitens: Sie erlaubt ihnen die Abgabe von Offerten. Somit trägt sie zu mehr Transparenz und Fairness im öffentlichen Beschaffungswesen bei.

#### Zielsetzung
Unsere Grundidee ist die Entwicklung einer künstlichen Intelligenz (KI), die als digitaler Berater eingesetzt werden kann. Diese KI soll Unternehmen, die sich über Simap.ch bewerben, eine strategische Analyse bieten. Genau gesagt wollen wir folgende Vorteile schaffen:

1️⃣ CPV – Common Procurement Vocabulary

→ EU-weite Klassifikation für öffentliche Ausschreibungen

Bedeutung: „Was wird beschafft?“

Einsatz auf simap.ch: Jede Ausschreibung hat mindestens einen CPV-Code, der die Art der Leistung beschreibt.

Beispiel: 45000000-7 = Bauarbeiten, 72000000-5 = IT-Dienstleistungen.

Ziel: Einheitliche Kategorisierung von Ausschreibungen in ganz Europa.

Nutzen für dich: Grundlage für Branchen-Matching, Textklassifikation, Feature Engineering.

🧱 2️⃣ BKP – Baukostenplan (Schweiz)

→ Klassifikationssystem für Bauprojekte (nach SIA / CRB)

Bedeutung: „Woraus bestehen die Baukosten?“

Struktur: Numerische Gliederung, z. B.

1: Grundstück,

2: Gebäude,

3: Betriebseinrichtungen,

4: Umgebung.

Beispiel: 221 = Tragstruktur Beton, 244 = Innenwände.

Einsatz: In simap.ch häufig bei Bauausschreibungen (z. B. Hochbau, Tiefbau), um Positionen zu strukturieren.

Nutzen für dich: Wenn du ML auf Bauprojekte machst → Feature für Art der Arbeiten / Kostenstruktur.

🧩 3️⃣ NPK – Normpositionen-Katalog

→ Standardisiertes Leistungsverzeichnis für Bauarbeiten (CH-Norm SIA 451)

Bedeutung: „Welche einzelnen Leistungen werden im Bau ausgeschrieben?“

Struktur: Katalog mit Tausenden Positionen (z. B. 221.111 = Betonwand einschalen und betonieren).

Einsatz auf simap.ch: In Bauausschreibungen als Referenz für Leistungsverzeichnisse.

Ziel: Vergleichbare, normierte Angebotspositionen → alle Anbieter bieten auf die gleichen Leistungen.

Nutzen für dich: Sehr detaillierte Featurequelle — du kannst Leistungstypen clustern oder analysieren, welche NPK-Gruppen höhere Zuschlagsraten haben.

🏛️ 4️⃣ OAG – Öffentliche Ausschreibung gemäss Gesetz

→ Rechtlicher Rahmenbegriff für Beschaffung nach BöB / VöB / IVöB

Bedeutung: Verfahrenstyp oder gesetzliche Grundlage der Ausschreibung.

Beispiel: „OAG-Verfahren“ = öffentliches, offenes Verfahren (alle dürfen bieten).

In simap.ch: Wird im Metadatenbereich der Ausschreibung oder als Filter verwendet (z. B. OAG = offenes Verfahren, Einladungsverfahren, freihändig).

Nutzen für dich: Klares Feature für dein ML-Modell — die Zuschlagschancen unterscheiden sich je nach Verfahrensart stark.

🏗️ 5️⃣ eBKP-H – Elementbasierter Baukostenplan (Hochbau)

→ Modernere, digital strukturierte Version des BKP speziell für Hochbau

Bedeutung: „Welche Bauelemente werden erstellt?“ (nicht nur Kostenstellen)

Beispiel:

D200 = Aussenwände,

D300 = Decken,

D400 = Dach.

Einsatz: In digitalen Ausschreibungen oder BIM-Projekten (Building Information Modeling).

Nutzen für dich: Wenn du Bauprojekte mit KI analysierst, liefert eBKP-H semantisch klarere Struktur als BKP → bessere Features für Baukategorien.

🚧 6️⃣ eBKP-T – Elementbasierter Baukostenplan (Tiefbau)

→ Pendant zu eBKP-H, aber für Tiefbau (Infrastruktur, Strassen, Wasser, Energie)

Bedeutung: Strukturierung nach Bauobjekten des Tiefbaus.

Beispiel:

C100 = Erdarbeiten,

C300 = Strassenbau,

C500 = Entwässerungssysteme.

Einsatz: In simap.ch bei Tiefbauausschreibungen, wenn der Auftraggeber digital plant oder Leistungsverzeichnisse gemäss CRB-Standard publiziert.

Nutzen für dich: Gleiches Prinzip wie eBKP-H → strukturierte Klassifikation von Tiefbauleistungen für Analysen.
