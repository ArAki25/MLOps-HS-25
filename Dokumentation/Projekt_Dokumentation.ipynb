## Ziele und Aufgabenstellung

### Das Projektkonzept

Simap.ch ist die offizielle Schweizer Publikationsplattform für öffentliche Beschaffungen. Diese Plattform hat eine zentrale Funktion: Sie ermöglicht Unternehmen Einsicht in Ausschreibungen und erlaubt ihnen die Abgabe von Offerten. Somit trägt sie zu mehr Transparenz und Fairness im öffentlichen Beschaffungswesen bei.

### Zielsetzung

Unsere Grundidee ist die Entwicklung einer künstlichen Intelligenz (KI), die als digitaler Berater eingesetzt werden kann. Diese KI soll Unternehmen, die sich über simap.ch bewerben, eine strategische Analyse bieten. Konkret wollen wir folgende Vorteile schaffen:

- **Projektbewertung**: Identifizierung von Projekten, die für die jeweilige Firma sinnvoll und erfolgsversprechend sind.
- **Zuschlagsanalyse**: Bewertung der Erfolgschancen basierend auf historischen Zuschlägen und Mustern (Optional).

### Erwartete Ergebnisse

Am Ende des Projekts soll eine funktionsfähige Webapplikation entstehen, die:
- Alle aktuellen Ausschreibungen von simap.ch anzeigt
- Such- und Filterfunktionen bietet
- Für Pro-Benutzer personalisierte ML-basierte Empfehlungen liefert

## Theorie

### Öffentliche Beschaffung in der Schweiz

Das öffentliche Beschaffungswesen in der Schweiz unterliegt strengen gesetzlichen Regelungen. Öffentliche Auftraggeber wie Bund, Kantone und Gemeinden sind verpflichtet, Aufträge ab einem bestimmten Schwellenwert öffentlich auszuschreiben. Dies gewährleistet Transparenz und Wettbewerb.

### Machine Learning für Empfehlungssysteme

Für die Klassifikation von Ausschreibungen verwenden wir überwachtes Lernen. Das Modell wird mit historischen Daten trainiert, bei denen bekannt ist, welche Ausschreibungen für bestimmte Firmentypen relevant waren. Basierend auf Merkmalen wie Auftragstyp, Region, CPV-Code und Beschreibungstext kann das Modell vorhersagen, ob eine neue Ausschreibung für eine Firma interessant ist.

### Webentwicklung mit Flask

Flask ist ein leichtgewichtiges Python-Web-Framework, das sich besonders für kleinere bis mittlere Webanwendungen eignet. Es folgt dem WSGI-Standard und ermöglicht eine schnelle Entwicklung von Web-APIs und Benutzeroberflächen.

## Methodik, Vorgehensweise und Prozeduren

### API-Anbindung und Datenbeschaffung

Damit wir unsere Klassifikationsaufgabe durchführen können, brauchen wir strukturierte historische Daten zu Ausschreibungen und Zuschlägen von simap.ch. Weil simap.ch im Web-Interface keine offenen Download-Funktionen anbietet, haben wir entschieden, die offizielle REST-API von simap.ch zu verwenden. Damit beziehen wir die Daten programmgesteuert und exportieren sie in ein geeignetes Analyseformat.

#### API von simap.ch einrichten

Folgende Schritte wurden durchgeführt:

1. Registrierung auf simap.ch als Organisation, um Zugriff auf die neue Plattform (ab 2024) zu erhalten.
2. Anfrage nach API-Zugang über die Administratoren von simap.ch.
3. Nach Freischaltung erhielten wir einen API-Key (Bearer Token) zur Authentifizierung, die Basis-URL der Schnittstelle sowie die API-Dokumentation mit verfügbaren Endpunkten.

#### Datenexport und Aufbereitung

Um die Daten weiter in Python analysieren zu können, haben wir die abgerufenen JSON-Daten in eine CSV-Datei umgewandelt. Anschliessend wurden relevante Attribute für unsere Klassifikationsmodelle ausgewählt.

Für die Klassifikation haben wir uns für eine binäre Zielvariable entschieden:
- **ist_interessant**: Die Ausschreibung ist für die Firma relevant
- **nicht_interessant**: Die Ausschreibung ist nicht relevant

Diese Einteilung basiert auf verschiedenen Kriterien wie Höhe des Auftragswerts, Projekttyp, Region und textliche Beschreibung des Projekts.

### Datenbank-Integration mit Supabase

Für die Speicherung, Verwaltung und Bereitstellung der strukturierten Projektdaten haben wir Supabase als Datenbanklösung eingesetzt. Supabase bietet eine PostgreSQL-basierte Cloud-Datenbank mit integrierten Authentifizierungs- und API-Funktionen.

Dies ermöglichte uns:
- Eine sichere Speicherung der Ausschreibungsdaten
- Eine klare Strukturierung der Tabellen (Ausschreibungen, Firmen, Klassifikationsergebnisse)
- Eine einfache Anbindung der Webapplikation über REST-Endpunkte
- Role-based Access für einzelne Benutzer

Supabase fungiert als zentrale Datendrehscheibe, welche sowohl von der Python-Webapplikation als auch vom ML-Modell angesprochen wird.

### Webentwicklung

Die Webapplikation wurde mit Python Flask entwickelt. Der Entwicklungsprozess umfasste:

1. **Backend-Entwicklung**: Erstellung der Flask-Routen und API-Endpoints
2. **Datenbankanbindung**: Implementierung der Supabase-Client-Funktionen
3. **Frontend-Entwicklung**: Gestaltung der HTML-Seiten mit CSS und JavaScript
4. **Deployment**: Veröffentlichung auf Render.com

## Ergebnisse

### Webapplikation: SAJF Strategies

Wir haben eine B2B Web-Applikation entwickelt, die Schweizer Unternehmen hilft, öffentliche Ausschreibungen von simap.ch zu finden. Die Plattform bietet eine kostenlose Version mit Such- und Filterfunktionen sowie eine Pro-Version mit ML-basierten Empfehlungen.

### Verwendete Technologien

| Komponente | Technologie | Begründung |
|------------|-------------|------------|
| Backend | Python Flask | Einfach zu lernen, gut für kleinere Webanwendungen |
| Datenbank | Supabase (PostgreSQL) | Gehostete Datenbank mit eingebauter API |
| Frontend | HTML, CSS, JavaScript | Keine zusätzlichen Frameworks nötig |
| Hosting | Render.com | Kostenloses Hosting, GitHub-Integration |

### Projektstruktur

Die Struktur der Webseite besteht aus mehreren Dateien:

- **app.py**: Hauptdatei, welche den Flask-Server startet und alle Routen definiert
- **supabase_client.py**: Enthält alle Funktionen für die Datenbankverbindung und Abfragen
- **requirements.txt**: Liste aller Python-Pakete, die installiert werden müssen
- **templates/**: Ordner mit allen HTML-Seiten (Hauptseite, Login-Seite, Pro-Dashboard)

### Supabase-Einrichtung

Zuerst haben wir ein Projekt auf supabase.com erstellt. Dabei erhält man eine Project URL und einen API Key, die später für die Verbindung benötigt werden.

In Supabase haben wir mehrere Tabellen erstellt:

| Tabelle | Beschreibung |
|---------|--------------|
| projects_website | Alle Ausschreibungen von simap.ch |
| pro_users | Login-Daten der Pro-Benutzer |
| BauUnternehmenAG | ML-Predictions für Beispiel-Baufirma |
| ITAG | ML-Predictions für Beispiel-IT-Firma |
| BeschaffungsAG | ML-Predictions für Beispiel-Beschaffungsfirma |

Wichtig war, dass wir die Row Level Security für alle Tabellen deaktiviert haben, damit die Website ohne Authentifizierung auf die Daten zugreifen kann.

### Render.com Deployment

Für das Hosting haben wir Render.com verwendet. Der Einrichtungsprozess umfasste:

1. Anmeldung auf Render.com und Erstellung eines neuen Web Service
2. Verbindung mit dem GitHub-Repository
3. Konfiguration der Build-Einstellungen (Python 3, pip install, gunicorn)
4. Setzen der Umgebungsvariablen (SUPABASE_URL, SUPABASE_KEY, SECRET_KEY)

Nach dem Einrichten wird bei jedem Git Push automatisch eine neue Version deployed. Der Prozess dauert etwa zwei bis drei Minuten.

### Architektur und Komponentenverbindung

Die Website verbindet drei Hauptkomponenten miteinander:
```
simap.ch (Daten) → Supabase (Datenbank) ← ML-Modelle (Predictions)
                          ↓
                   Flask Website (Render.com)
                          ↓
                      Benutzer
```

Die Ausschreibungsdaten kommen ursprünglich von simap.ch und werden in der Supabase-Datenbank gespeichert. Das vortrainierte ML-Modell sagt vorher, welche Aufträge für welche Firma relevant sind. Diese Predictions werden ebenfalls in Supabase gespeichert.

Die Flask-Website läuft auf Render.com und verbindet sich mit Supabase über API-Aufrufe. Wenn ein Benutzer die Seite öffnet, holt die Website die Daten aus der Datenbank und zeigt sie an. Bei der Pro-Version werden zusätzlich die ML-Predictions aus den firmenspezifischen Tabellen geladen.

### Website-Features

#### Kostenlose Version

- **Suchfunktion**: Suche nach Keywords in Titel und Beschreibung
- **Kanton-Filter**: Alle 26 Schweizer Kantone auswählbar
- **Verfahrenstyp-Filter**: Offen, Selektiv, Einladung
- **Auftragstyp-Filter**: Bau, Dienstleistung, Lieferung
- **Detailansicht**: Klick auf Karte zeigt alle Details
- **Direktlinks**: Jeder Auftrag verlinkt zur offiziellen Ausschreibung auf simap.ch

#### Pro-Version

- **Login-System**: Firmen können sich mit Email und Passwort einloggen
- **Personalisierte Empfehlungen**: ML-basierte Auftragsvorschläge
- **Match-Rate**: Prozentuale Angabe der Relevanz (z.B. 95% Match)
- **Sortierung**: Aufträge nach Relevanz sortiert

### API-Endpoints

| Endpoint | Beschreibung |
|----------|--------------|
| /api/projects | Alle Projekte (limit 200) |
| /api/search?q=keyword | Suche nach Keyword |
| /api/filter?canton=ZH | Filter nach Kanton |
| /api/cantons | Liste aller Kantone mit Anzahl |
| /api/statistics | Total und heute neu |

## Diskussion und Interpretation

### Erreichte Ziele

Die Hauptziele des Projekts wurden erfolgreich erreicht. Die Webapplikation ist funktionsfähig und bietet sowohl eine kostenlose als auch eine Pro-Version an. Die Integration der ML-Modelle funktioniert und liefert sinnvolle Empfehlungen für die Beispielfirmen.

### Stärken der Lösung

- **Benutzerfreundlichkeit**: Die Oberfläche ist intuitiv und übersichtlich gestaltet
- **Skalierbarkeit**: Die Architektur erlaubt einfache Erweiterungen
- **Aktualität**: Daten werden regelmässig von simap.ch synchronisiert
- **Personalisierung**: Pro-Benutzer erhalten massgeschneiderte Empfehlungen

### Limitationen

- **Datenmenge für ML**: Die Trainingsdaten sind begrenzt, was die Genauigkeit der Vorhersagen einschränkt
- **Hosting-Einschränkungen**: Der kostenlose Render-Tarif hat Performance-Limitationen
- **Manuelle Firmenprofile**: Neue Firmen müssen manuell angelegt und trainiert werden

### Zusammenarbeit im Team

Die Zusammenarbeit in unserem Team war erneut sehr produktiv. Wir haben früh Rollen und Aufgaben definiert und die Arbeit klar verteilt:

- **Akishan**: Kommunikation mit simap.ch, API-Zugänge, Training des ML-Modells
- **Saliou**: Datenaufbereitung und Einrichtung der Datenbank
- **Janarthanan und Fenlin**: Frontend/Website-Erstellung und Verbindung der Komponenten

#### Organisation und Meetings

- Wöchentliche Treffen im Rahmen des Unterrichts
- Zusätzliche Treffen ausserhalb der Schule (online oder vor Ort)
- Gegenseitige Deadlines wurden gesetzt und eingehalten
- Aufgaben wurden über GitHub verwaltet, sodass jede Änderung nachvollziehbar war

## Schlussfolgerungen

Das Projekt hat gezeigt, dass es möglich ist, mit relativ einfachen Mitteln eine funktionsfähige B2B-Plattform zu entwickeln. Die Kombination aus Flask, Supabase und Render.com ermöglicht eine schnelle Entwicklung und Bereitstellung von Webanwendungen.

Die Integration von Machine Learning in die Plattform bietet einen echten Mehrwert für Unternehmen, die regelmässig an öffentlichen Ausschreibungen teilnehmen. Durch die personalisierten Empfehlungen können sie Zeit sparen und sich auf die relevantesten Aufträge konzentrieren.

### Herausforderungen

Während des Projekts mussten wir uns verschiedenen Herausforderungen stellen:

#### Technische Hürden bei der API-Nutzung
Der Kontakt mit den Verantwortlichen von simap.ch hat mehr Zeit in Anspruch genommen als geplant. Die API ist nicht öffentlich zugänglich, daher war es notwendig, persönliche Freigaben und Registrierungen zu beantragen.

#### Häufige Probleme der Webseite
- Application Errors durch fehlende Umgebungsvariablen
- Keine Datenanzeige wegen aktivierter Row Level Security
- Langsame Ladezeiten durch den kostenlosen Render-Tarif (Schlafmodus nach 15 Minuten Inaktivität)
